### 1. Objectif principal

NumPy : Focalisé sur le calcul numérique. Il est conçu pour manipuler des tableaux (arrays) et matrices de données numériques (entiers, flottants) de manière efficace et vectorisée. C'est une bibliothèque de base pour le traitement de données massives, l'algèbre linéaire et les opérations rapides.

## Exercice 2 — Commutant de A = $ \begin{pmatrix}1 & 1 \\ 0 & 1\end{pmatrix}$

Trouvez les matrices B telles que AB = BA.

Dans le code, vous pouvez définir B générique et calculer AB - BA et voir les conditions.

```python
A = sp.Matrix([[1, 1], [0, 1]])

# vous pouvez supposer des valeurs ou utiliser sympy pour symbolique.

a, b, c, d = sp.symbols('a b c d')
B_sym = sp.Matrix([[a, b], [c, d]])
A_sym = sp.Matrix([[1, 1], [0, 1]])
B_sym = sp.Matrix([[a, b], [c, d]])
```

```python
comm = A_sym*B_sym - B_sym*A_sym
print('AB - BA =')
comm 
```

```python
# Extraire les équations : chaque élément de comm == 0
equations = []
for i in range(comm.shape[0]):  # Lignes
    for j in range(comm.shape[1]):  # Colonnes
        if comm[i, j] != 0:  # Ignorer les éléments déjà nuls
            equations.append(sp.Eq(comm[i, j], 0))
```

```python
equations
```

```python
# Résoudre le système pour les variables a, b, c, d
solutions = sp.solve(equations, [a, b, c, d])

# Afficher les solutions
print('Solutions :')
sp.pprint(solutions)
```

$AB-BA=\begin{pmatrix}c&-a+d\\0&-c\end{pmatrix}=0
\;\Longrightarrow\; c=0,\ d=a.$

Ainsi
$$
\boxed{B=\begin{pmatrix}\alpha&\beta\\[2pt]0&\alpha\end{pmatrix},\ \alpha,\beta\in\mathbb R.}
$$
(Le commutant est l’algèbre des polynômes en $A$, i.e. $\alpha I+\beta(A-I)$.)

## Exercice 9 — Binôme

A = $\begin{pmatrix}1 & 1 & 0 \\ 0 & 1 & 1 \\ 0 & 0 & 1\end{pmatrix}$

B = A - I

Calculez puissances de B, puis de A.

```python
I = np.eye(3)
A = sp.Matrix([[1,1,0],[0,1,1],[0,0,1]])
B = A - I

B2 = B**2
print('B2 =', B2)
B3 = B**3
print('B3 =', B3)  # Should be zero

# A^n = I + n B + binom(n,2) B2
from scipy.special import binom
n = 5
An = I + n * B + binom(n, 2) * B2
print('A^n =', An)

# Vérifiez en calculant directement
An_direct = np.linalg.matrix_power(A, n)
print('A^n direct =', An_direct)
```

## Exercice 10 — Puissance k-ième

U = matrice 4x4 avec 0 sur diagonale, 1 ailleurs.

J = matrice tout 1 4x4

U = J - I

Calculez U², etc.

```python
I4 = sp.eye(4)
J = sp.ones(4,4)
U = J - I4

U2 = U**2
print('U2 =', U2)

# Formule U^k = (3^k - (-1)^k)/4 * U + (3^k + 3*(-1)^k)/4 * I4
k = 2
coeff_U = (3**k - (-1)**k) / 4
coeff_I = (3**k + 3*(-1)**k) / 4
Uk = coeff_U * U + coeff_I * I4
Uk
```
